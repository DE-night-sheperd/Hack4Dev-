# 🏗️ Building Age Classification: Transferable UnitModel Architecture

## 1. Challenge & Vision: City-Agnostic Urban Intelligence
In the **Building Age Classification Hackathon**, our goal is to classify 30 m × 30 m Landsat pixels into **4 discrete historical building age classes** using multi-decadal satellite spectral observations (1984–2024).

The core difficulty is **cross-city domain shift**: Madrid and Amsterdam possess fundamentally different building materials (stucco/concrete vs. brick), urban density, and microclimates. A direct supervised model trained on Madrid collapses when evaluated on Amsterdam in zero-shot settings.

Our **Transferable UnitModel** addresses this through **Domain-Invariant Metric Learning & Prototypical Few-Shot Transfer**, learning what constitutes relative era differences rather than memorizing city-specific absolute reflectance.

---
## 2. Age Class Definitions
Buildings are categorized into 4 historical classes using the continuous :

| Class | Period (Madrid) | Period (Amsterdam) | Historical & Spectral Context |
|:-----:|:---------------:|:------------------:|:-----------------------------|
| **1** | Pre-1960        | Pre-1945           | Pre-modern traditional masonry; stable settled spectral footprint |
| **2** | 1960–1984       | 1945–1984          | Post-war expansion pre-dating Landsat satellite record |
| **3** | 1984–2004       | 1984–2004          | Modern expansion; construction events occur during early Landsat era |
| **4** | 2004–2024       | 2004–2024          | Contemporary era; high spectral variance and construction disruptions |

---
## 3. Four Core Pillars of the UnitModel

### A. Spectral-Spatial-Temporal Feature Suite (132 Dimensions)
- **Multi-Angle / Intra-Annual Variance:** Captures spectral variability across the 3 annual observations.
- **Atmospheric De-Noising:** Filters out undetected cirrus/haze anomalies (Blue > 15,000 threshold).
- **Spectral Indices:** NDVI (vegetation), NDBI (built-up), UI (urban index), MNDWI (water/moisture), BSI (bare soil).
- **Spatial Urban Fabric:** Real-space neighborhood context computed via KD-Tree on  grid, capturing local morphology and clustering without requiring external vector files.
- **Temporal Trajectories:** Early-period (1984–2003), Late-period (2004–2024), and Year-on-Year change rates.

### B. Domain-Adversarial Neural Network (DANN)
Uses a **Gradient Reversal Layer (GRL)** connected to a domain discriminator. While Madrid labeled samples optimize age classification, unlabeled Amsterdam samples are aligned with Madrid in the latent embedding space, stripping away city-specific spectral biases.

### C. Deep Metric Learning & Prototypical Few-Shot Transfer
Projects features into an **L2-normalized unit hypersphere embedding space** (16 dimensions). Transfer to Amsterdam is performed via **Prototypical Matching** across  samples per class. This eliminates catastrophic forgetting and achieves optimal Macro F1 even in extreme low-data regimes.

### D. Bayesian Uncertainty Quantification (MC-Dropout)
Applies Monte Carlo Dropout during inference over stochastic passes, outputting predictive mean confidence and entropy. This flags low-confidence or mixed-age pixels for active learning.

---
## 4. Evaluation Criteria & Competition Deliverables
1. **Madrid Repeated CV:** 5-fold cross-validation Macro F1.
2. **Amsterdam Few-Shot Adaptation:** Mean and standard deviation Macro F1 across 10 randomized trials for  samples/class.
3. **Target Prizes:**
   - **Best Low-Data F1 (25 samples/class)** (primary objective)
   - **Best Peak F1** & **Best Average F1**
   - **Best Originality** (domain-adversarial metric transfer)
