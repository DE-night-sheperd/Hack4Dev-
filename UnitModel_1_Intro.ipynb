## 3. The "Breathtaking" Components

### A. Temporal Transformer Encoder (The "Time Machine")
Instead of simple medians, we use a **Transformer-based attention mechanism** to scan the 40-year Landsat time-series. It automatically identifies the *exact year* of spectral change (construction/renovation) by attending to significant shifts in NIR, SWIR, and **Thermal** bands.

### B. Graph Neural Morphology (The "Urban Context")
Cities aren't just pixels; they are networks. We treat the city as a **Graph** where buildings are nodes. Using **Graph Convolutional Networks (GCN)**, the model learns urban morphology—how the age of one building is topologically linked to its neighbors, capturing how historic centers expand into modern suburbs.

### C. Bayesian Uncertainty (The "Honest Model")
Using **Monte Carlo Dropout**, our model provides a **Confidence Score** for every prediction. If it sees a building type it hasn't seen before in Singapore, it won't just guess; it will signal high uncertainty, allowing for smart Active Learning.

### D. PSO-Memetic Super-Optimizer
A hybrid metaheuristic that combines the global exploration of **Particle Swarm Optimization** with the surgical local refinement of **Memetic Search**, ensuring we squeeze every decimal point of F1-score out of the transfer process.
